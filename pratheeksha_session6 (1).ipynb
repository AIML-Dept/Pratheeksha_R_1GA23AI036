```python
# ============================================================
# SIMON'S ALGORITHM - SIMPLE LAB CODE
# Easy + Medium + Hard + Real World
# Run this whole cell in Google Colab
# ============================================================

!pip -q install qiskit qiskit-aer

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
import numpy as np


# ============================================================
# EASY
# Create a simple Simon function for secret = 110
# ============================================================

print("========== EASY ==========")

secret = "110"

def f(x):
    # Simple function satisfying f(x) = f(x XOR secret)
    return x[0] ^ x[2]

print("Truth Table:")
for x in range(8):
    bits = format(x, "03b")
    print(bits, "->", f([int(bits[0]), int(bits[1]), int(bits[2])]))

print("\nChecking Simon property:")

for x in range(8):
    y = x ^ int(secret, 2)

    xb = format(x, "03b")
    yb = format(y, "03b")

    fx = f([int(xb[0]), int(xb[1]), int(xb[2])])
    fy = f([int(yb[0]), int(yb[1]), int(yb[2])])

    print(xb, "and", yb, ":", fx == fy)


# ============================================================
# MEDIUM
# Build Simon's quantum circuit
# ============================================================

print("\n\n========== MEDIUM ==========")

secret = "110"
n = len(secret)

qc = QuantumCircuit(n, n)

# Put qubits into superposition
for i in range(n):
    qc.h(i)

# Simple oracle for secret 110
# f(x) = x0 XOR x2
qc.cx(0, 2)

# Hadamard again
for i in range(n):
    qc.h(i)

# Measure
qc.measure(range(n), range(n))

print(qc.draw())

simulator = AerSimulator()

result = simulator.run(
    qc,
    shots=1024
).result()

counts = result.get_counts()

print("Measurement results:")
print(counts)


# ============================================================
# HARD
# 4-bit example
# Secret = 1011
# Use measured equations to find the secret
# ============================================================

print("\n\n========== HARD ==========")

secret = "1011"

# Example equations obtained from Simon measurements
# y.s = 0 (mod 2)

equations = [
    "1001",
    "0110",
    "1100"
]

print("Secret is hidden.")
print("Measurement equations:")

for eq in equations:
    print(eq, " . s = 0")


# Try every possible 4-bit string
possible = []

for number in range(16):

    candidate = format(number, "04b")

    good = True

    for eq in equations:

        value = 0

        for i in range(4):
            value += int(eq[i]) * int(candidate[i])

        if value % 2 != 0:
            good = False

    if good:
        possible.append(candidate)


print("\nPossible solutions:")
print(possible)

# Remove 0000
non_zero = [x for x in possible if x != "0000"]

print("\nRecovered secret:")
print(non_zero[0] if len(non_zero) == 1 else non_zero)


# ============================================================
# REAL WORLD
# Compare classical and quantum queries
# ============================================================

print("\n\n========== REAL WORLD ==========")

print("Number of bits | Classical | Quantum")

for n in range(2, 11):

    classical = 2 ** n
    quantum = n

    print(
        n,
        "            |",
        classical,
        "       |",
        quantum
    )


# Simple plot
import matplotlib.pyplot as plt

n_values = range(2, 11)

classical = [2**n for n in n_values]
quantum = [n for n in n_values]

plt.plot(
    n_values,
    classical,
    marker="o",
    label="Classical"
)

plt.plot(
    n_values,
    quantum,
    marker="o",
    label="Quantum"
)

plt.xlabel("Number of bits")
plt.ylabel("Number of queries")
plt.title("Classical vs Quantum Queries")

plt.legend()
plt.grid()

plt.show()


# ============================================================
# FINAL
# ============================================================

print("\n========== DONE ==========")
print("Easy    -> Simon function checked")
print("Medium  -> Quantum circuit executed")
print("Hard    -> GF(2) equations solved")
print("Real    -> Classical vs Quantum comparison completed")
```
