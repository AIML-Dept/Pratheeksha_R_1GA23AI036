```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 1)

# Set the second qubit to |1>
qc.x(1)

# Apply Hadamard transformation
qc.h(0)
qc.h(1)

# Oracle for balanced function f(x) = x
qc.cx(0, 1)

# Apply Hadamard to the first qubit
qc.h(0)

# Store the measurement
qc.measure(0, 0)

print(qc.draw())

sim = AerSimulator()
simulation_result = sim.run(qc, shots=1024).result()

output_counts = simulation_result.get_counts()
print("Final output:", output_counts)
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 1)

# Prepare the target qubit
qc.x(1)

# Create superposition of both qubits
qc.h(0)
qc.h(1)

# Apply the oracle using a CNOT gate
qc.cx(0, 1)

# Transform the input qubit
qc.h(0)

# Measure the input
qc.measure(0, 0)

print(qc.draw())

sim = AerSimulator()
job = sim.run(qc, shots=1024)
result = job.result()

counts = result.get_counts()
print("Measurement output:", counts)
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def run_deutsch(oracle_type):
    circuit = QuantumCircuit(2, 1)

    # Initialize the ancilla qubit
    circuit.x(1)

    # Apply Hadamard gates
    circuit.h(0)
    circuit.h(1)

    # Select the oracle
    if oracle_type == "zero":
        pass
    elif oracle_type == "one":
        circuit.x(1)
    elif oracle_type == "identity":
        circuit.cx(0, 1)
    elif oracle_type == "not":
        circuit.x(1)
        circuit.cx(0, 1)
    else:
        raise ValueError("Unknown oracle type")

    # Final transformation and measurement
    circuit.h(0)
    circuit.measure(0, 0)

    return circuit


backend = AerSimulator()

oracle_list = ["zero", "one", "identity", "not"]

for oracle in oracle_list:
    circuit = run_deutsch(oracle)
    simulation = backend.run(circuit, shots=1024).result()

    print("\nOracle type:", oracle)
    print("Measurement counts:", simulation.get_counts())
    print(circuit.draw())
```
