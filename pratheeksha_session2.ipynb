```python
import sys
!{sys.executable} -m pip install qiskit

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import random

# Function to get and print the statevector
def show_result(name, circuit):
    state = Statevector.from_instruction(circuit)

    print("\n" + "-" * 55)
    print(name)
    print("-" * 55)
    print(circuit.draw())
    print("Statevector:", state)

    return state


# =====================================================
# EXERCISE 1 : Pauli X, Y and Z Gates
# =====================================================
print("\nEXERCISE 1 : PAULI GATES")

gates = {
    "Pauli-X": "x",
    "Pauli-Y": "y",
    "Pauli-Z": "z"
}

for name, gate in gates.items():

    circuit = QuantumCircuit(1)

    if gate == "x":
        circuit.x(0)
    elif gate == "y":
        circuit.y(0)
    else:
        circuit.z(0)

    show_result(name, circuit)


# =====================================================
# EXERCISE 2 : Construct |-> State
# =====================================================
print("\nEXERCISE 2 : CONSTRUCT |-> STATE")

minus_state = QuantumCircuit(1)

# First create |+>
minus_state.h(0)

# Apply Z to obtain |->
minus_state.z(0)

show_result("State |-> using H and Z", minus_state)


# =====================================================
# EXERCISE 3 : Random Sequence of Five Gates
# =====================================================
print("\nEXERCISE 3 : RANDOM FIVE-GATE CIRCUIT")

available_gates = ["H", "X", "Y", "Z", "S", "T"]

random_gates = random.sample(available_gates, 5)

print("Selected Gates:", random_gates)

random_circuit = QuantumCircuit(1)

for gate in random_gates:

    if gate == "H":
        random_circuit.h(0)

    elif gate == "X":
        random_circuit.x(0)

    elif gate == "Y":
        random_circuit.y(0)

    elif gate == "Z":
        random_circuit.z(0)

    elif gate == "S":
        random_circuit.s(0)

    elif gate == "T":
        random_circuit.t(0)

show_result("Random Five-Gate Circuit", random_circuit)


# =====================================================
# EXERCISE 4 : Quantum Coin Flip
# =====================================================
print("\nEXERCISE 4 : QUANTUM COIN FLIP")

coin_circuit = QuantumCircuit(1)

# Hadamard creates equal superposition
coin_circuit.h(0)

coin_state = Statevector.from_instruction(coin_circuit)

print("Circuit:")
print(coin_circuit.draw())

print("\nProbability of Heads (0):",
      coin_state.probabilities_dict().get('0', 0))

print("Probability of Tails (1):",
      coin_state.probabilities_dict().get('1', 0))


# =====================================================
# EXERCISE 5 : Verify HZH = X
# =====================================================
print("\nEXERCISE 5 : VERIFY HZH = X")

# Create HZH circuit
hzh = QuantumCircuit(1)
hzh.h(0)
hzh.z(0)
hzh.h(0)

# Create X circuit
x_gate = QuantumCircuit(1)
x_gate.x(0)

# Get statevectors
hzh_state = Statevector.from_instruction(hzh)
x_state = Statevector.from_instruction(x_gate)

print("\nHZH Circuit:")
print(hzh.draw())

print("HZH Statevector:")
print(hzh_state)

print("\nX Gate Circuit:")
print(x_gate.draw())

print("X Statevector:")
print(x_state)

# Check whether both circuits are equivalent
if hzh_state.equiv(x_state):
    print("\nResult: HZH is equivalent to X.")
else:
    print("\nResult: HZH is NOT equivalent to X.")
```
