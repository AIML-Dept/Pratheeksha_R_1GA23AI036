```python
# ==========================================================
# Tutorial 2
# Qubits, State Vectors and Superposition Visualization
# ==========================================================

# Uncomment if Qiskit is not installed
# !pip install qiskit qiskit-aer matplotlib pylatexenc

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector, plot_state_qsphere
import numpy as np
import matplotlib.pyplot as plt

print("="*70)
print("Tutorial 2: Qubits, State Vectors and Superposition Visualization")
print("="*70)

# ==========================================================
# Exercise 1
# Create |0>, Apply X Gate
# ==========================================================

print("\n")
print("="*70)
print("Exercise 1 : X Gate")
print("="*70)

qc1 = QuantumCircuit(1)

qc1.x(0)

state1 = Statevector.from_instruction(qc1)

print(qc1.draw())

print("\nStatevector")
print(state1)

print("\nProbabilities")
print(state1.probabilities())

plot_bloch_multivector(state1)
plt.show()

# ==========================================================
# Exercise 2
# Hadamard + S Gate
# ==========================================================

print("\n")
print("="*70)
print("Exercise 2 : Hadamard + S Gate")
print("="*70)

qc2 = QuantumCircuit(1)

qc2.h(0)
qc2.s(0)

state2 = Statevector.from_instruction(qc2)

print(qc2.draw())

print("\nStatevector")
print(state2)

print("\nProbabilities")
print(state2.probabilities())

plot_bloch_multivector(state2)
plt.show()

# ==========================================================
# Exercise 3
# Equal Superposition of 2 Qubits
# ==========================================================

print("\n")
print("="*70)
print("Exercise 3 : Two-Qubit Superposition")
print("="*70)

qc3 = QuantumCircuit(2)

qc3.h(0)
qc3.h(1)

state3 = Statevector.from_instruction(qc3)

print(qc3.draw())

print("\nStatevector")
print(state3)

print("\nProbabilities")

labels = ['00','01','10','11']
probs = state3.probabilities()

for label,prob in zip(labels,probs):
    print(label,"=",prob)

print("\nSum of probabilities =",np.sum(probs))

plot_state_qsphere(state3)
plt.show()

# ==========================================================
# Exercise 4
# Random Phase Error
# ==========================================================

print("\n")
print("="*70)
print("Exercise 4 : Random Phase Error")
print("="*70)

theta = np.random.uniform(0,2*np.pi)

qc4 = QuantumCircuit(1)

qc4.h(0)

qc4.p(theta,0)

state4 = Statevector.from_instruction(qc4)

print("Random Phase =",theta)

print(qc4.draw())

print("\nStatevector")
print(state4)

print("\nProbabilities")
print(state4.probabilities())

plot_bloch_multivector(state4)
plt.show()

# ==========================================================
# Exercise 5
# Arbitrary Rotation
# ==========================================================

print("\n")
print("="*70)
print("Exercise 5 : Arbitrary State")
print("="*70)

def arbitrary_state(theta,phi):

    qc = QuantumCircuit(1)

    qc.ry(theta,0)
    qc.rz(phi,0)

    state = Statevector.from_instruction(qc)

    print(qc.draw())

    print("\nStatevector")
    print(state)

    plot_bloch_multivector(state)
    plt.show()

    return state

theta = np.pi/3
phi = np.pi/4

state5 = arbitrary_state(theta,phi)

# ==========================================================
# Manual Verification
# ==========================================================

print("\n")
print("="*70)
print("Manual Verification")
print("="*70)

alpha = np.cos(theta/2)
beta = np.exp(1j*phi)*np.sin(theta/2)

print("Alpha =",alpha)
print("Beta =",beta)

print("\nNormalization")

print("|alpha|² =",abs(alpha)**2)
print("|beta|² =",abs(beta)**2)

print("Sum =",abs(alpha)**2+abs(beta)**2)

# ==========================================================
# Reflection Answers
# ==========================================================

print("\n")
print("="*70)
print("Reflection Question 1")
print("="*70)

print("""
Why must the sum of squared amplitudes always equal 1?

Answer:
The squared magnitudes of the amplitudes represent the probabilities of
all possible measurement outcomes. Since one of the outcomes must occur,
their total probability must always equal 1. This is called normalization.
""")

print("="*70)
print("Reflection Question 2")
print("="*70)

print("""
How does the Bloch sphere fail to represent multi-qubit entangled states?

Answer:
The Bloch sphere can only represent a single qubit.
Entangled states involve correlations between multiple qubits and require
higher-dimensional representations such as the QSphere or Statevector.
""")

print("\n")
print("="*70)
print("Tutorial Completed Successfully")
print("="*70)
```
