```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

# Apply Hadamard gate to qubit 0
qc.h(0)

# Apply controlled-X gate
qc.cx(control_qubit=0, target_qubit=1)

# Store measurement results
qc.measure_all()

print(qc.draw())

sim = AerSimulator()
job = sim.run(qc, shots=1024)
result = job.result()

print("Final counts:", result.get_counts())
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

def generate_state(state_name):
    qc = QuantumCircuit(2, 2)

    # Prepare the entangled pair
    qc.h(0)
    qc.cx(0, 1)

    # Apply additional gates
    gates = {
        "Phi+": [],
        "Phi-": ["z"],
        "Psi+": ["x"],
        "Psi-": ["x", "z"]
    }

    for gate in gates[state_name]:
        if gate == "x":
            qc.x(0)
        elif gate == "z":
            qc.z(0)

    qc.measure_all()
    return qc


sim = AerSimulator()

for state in ["Phi+", "Phi-", "Psi+", "Psi-"]:
    circuit = generate_state(state)
    result = sim.run(circuit, shots=1024).result()

    print(state, ":", result.get_counts())
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

# Create an entangled pair
qc.h(0)
qc.cx(0, 1)

# Apply basis transformation
qc.h(1)
qc.h(0)

# Record the measurement outcomes
qc.measure(0, 0)
qc.measure(1, 1)

print(qc.draw())

sim = AerSimulator()
job = sim.run(qc, shots=1024)
result = job.result()

counts = result.get_counts()
print("Final output:", counts)
```
