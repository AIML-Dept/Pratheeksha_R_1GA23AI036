```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

# Generate entangled qubits
qc.h(0)
qc.cx(0, 1)

# Message selected by Alice: 00
# Identity encoding

# Decode the received message
qc.h(0)
qc.cx(0, 1)

# Store measurement results
qc.measure(0, 0)
qc.measure(1, 1)

print(qc.draw())

sim = AerSimulator()
job = sim.run(qc, shots=1024)
result = job.result()

print("Measurement counts:", result.get_counts())
```

```python

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

# Generate an entangled state
qc.h(0)
qc.cx(0, 1)

# Original message is 00
# Apply bit-flip noise to the first qubit
qc.x(0)

# Recover the classical message
qc.h(0)
qc.cx(0, 1)

# Record output bits
qc.measure(0, 0)
qc.measure(1, 1)

print(qc.draw())

simulator = AerSimulator()
job = simulator.run(qc, shots=1024)
result = job.result()

counts = result.get_counts()
print("Decoded output:", counts)
```
